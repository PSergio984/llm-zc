# Embeddings — LLM Zoomcamp Lesson 02

Using sentence-transformers with `all-MiniLM-L6-v2` to embed text and compute cosine similarity.

```python
from embedder import Embedder

embedder = Embedder()
```

## Encode a query

```python
q1 = "Can I still join the course after the start date?"
v1 = embedder.encode(q1)
print(f"Vector shape: {v1.shape}")
print(f"First 10 values: {v1[:10]}")
```

## Encode a relevant document

```python
d = "You don't need to register. You're accepted. You can also just start learning and submitting homework without registering."
dv = embedder.encode(d)

score = embedder.dot_product(v1, dv)
print(f"Q1 vs relevant doc: {score:.4f}")
```

## Encode an unrelated query

```python
q2 = "How to install Docker on Windows?"
v2 = embedder.encode(q2)

score = embedder.dot_product(v2, dv)
print(f"Q2 vs same doc: {score:.4f}")
```

The first score should be higher — Q1 (course registration) is semantically related to the document, while Q2 (Docker) is not.

## Cosine similarity via dot product

Because `all-MiniLM-L6-v2` outputs normalized vectors, `dot_product()` equals cosine similarity.

```python
cos_sim = embedder.similarity(v1, dv)
print(f"Cosine similarity: {cos_sim.item():.4f}")
print(f"Matches dot product: {abs(cos_sim.item() - embedder.dot_product(v1, dv)) < 1e-6}")
```
